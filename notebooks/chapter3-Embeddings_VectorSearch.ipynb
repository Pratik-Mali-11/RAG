# Lesson 3 — Embeddings and Vector Search
## 1. First, remember where we are

We currently have:

PDF
 ↓
Extract text
 ↓
Clean text
 ↓
Split into chunks

For example, our PDF might produce:

Chunk 1:
"Employees are entitled to 20 days of annual leave."


Chunk 2:
"Employees can carry forward up to 10 unused leave days."


Chunk 3:
"Medical insurance becomes available after 90 days."


Chunk 4:
"Employees must submit insurance claims within 30 days."

Now we have a problem:

How do we find the right chunk when the user asks a question?

Suppose the user asks:

"How many leave days can I carry forward?"

We need to find:

Chunk 2

This is where embeddings come in.

## 2. What is an Embedding?

The simplest explanation is:

An embedding converts text into numbers in such a way that text with similar meaning gets similar numbers.

For example:

"How many leave days can I carry forward?"

gets converted into something like:

[0.21, -0.73, 0.14, 0.89, ...]

Maybe 384 or 768 numbers depending on the embedding model.

Don't worry about the actual numbers.

The important thing is:

Text
 ↓
Embedding Model
 ↓
Vector

## 3. Why do we need numbers?

Computers are very good at comparing numbers.

They're not directly good at comparing the meaning of:

"How much vacation can I carry forward?"

and

"Employees may carry forward up to 10 unused leave days."

But after converting them into vectors, we can calculate how close they are.

Think of it like putting sentences on a map.

                 Medical insurance
                       ●

                      

     Leave policy
         ●
        ●
     ●
Question ●


                          Salary
                            ●


Things with similar meanings appear closer together.

## 4. A Simple Example

Imagine our embedding system only used two numbers.

We might get:

"How much vacation can I carry forward?"
        
        ↓

[0.90, 0.80]


"Employees can carry forward 10 leave days."
        
        ↓

[0.88, 0.82]


"Medical insurance starts after 90 days."
        
        ↓

[0.10, 0.20]

The first two vectors are close.

The third is far away.

Therefore:

Question
   
   ↓
   
[0.90, 0.80]


        ↓ similarity


[0.88, 0.82]
   Chunk 2

So we retrieve Chunk 2.
